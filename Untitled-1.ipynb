```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("shortform_video_engagement_3000.csv")

# Clean column names
df.columns = df.columns.str.strip()

# Use a real categorical column
counts = df['category'].value_counts()

# Keep top 4 categories
top = counts[:4]

# Combine remaining as "Others"
others = counts[4:].sum()

balanced = pd.concat([top, pd.Series({'Others': others})])

# Plot pie chart
plt.figure(figsize=(6,6))
balanced.plot(kind='pie', autopct='%1.1f%%', startangle=90)

plt.title("Video Category Distribution (Top 4 + Others)")
plt.ylabel("")
plt.axis('equal')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("shortform_video_engagement_3000.csv")

# Clean column names
df.columns = df.columns.str.strip()

# Target column from your dataset
target_col = 'is_trending'

# Split classes
df_majority = df[df[target_col] == 0]
df_minority = df[df[target_col] == 1]

# Oversample minority class
df_minority_upsampled = df_minority.sample(
    n=len(df_majority),
    replace=True,
    random_state=42
)

# Balanced dataset
df_balanced = pd.concat([df_majority, df_minority_upsampled], axis=0)

# Shuffle
df_balanced = df_balanced.sample(frac=1, random_state=42).reset_index(drop=True)

# Value counts
counts = df_balanced[target_col].value_counts()

# Pie chart
plt.figure(figsize=(6, 6))
plt.pie(
    counts.values,
    labels=['Not Trending (0)', 'Trending (1)'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Balanced Pie Chart (Is Trending)')
plt.axis('equal')
plt.show()
```
